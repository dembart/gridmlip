# Interactive view of energy maps

Hover an atom for its label, click two or three atoms for a distance or an angle, click a polyhedron for its bonds and volume; change the level, supercell, layers and theme in the panel.

### Default view: 1D percolation isosurface

```python
from ase.io import read
import gridmlip as gm

atoms = read("data/Li10Ge(PS6)2_mp.cif")
emap = gm.bvse(atoms, "Li+", resolution=0.2)
emap.view()
```

### 3D percolation level in a supercell

```python
b = emap.percolation_barriers()
emap.view(level=b["e3d"] + 0.01, supercell=(2, 2, 1))
```

### Ball-and-stick, light theme, looking down c

```python
emap.view(polyhedra=False, theme="light", along="c")
```

### Polyhedra only, rotation locked to c

```python
emap.view(bonds=False, isosurface=False, hide=["Li"], rotate_about="c")
```

### Save or open in a browser

```python
v = emap.view()
v.save("lgps_li.html")
# v.open()
```
