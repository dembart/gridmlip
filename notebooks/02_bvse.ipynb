### BVSE map

Oxidation states are assigned with pymatgen's BVAnalyzer unless `atoms.arrays["oxi_states"]` or `oxidation_states=` is given.

```python
from ase.io import read
import gridmlip as gm

atoms = read("data/Li10Ge(PS6)2_mp.cif")
emap = gm.bvse(atoms, "Li+", resolution=0.2)
emap.info["oxidation_states"]
```

### Percolation barriers

```python
emap.percolation_barriers()
```

### Li sites (local minima)

```python
emap.local_minima(min_distance=1.5, energy_window=0.5)[:5]
```

### Save for VESTA

```python
emap.write("lgps_bvse.grd")
```
