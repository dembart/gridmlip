# Training dataset for an MLIP

Defects → SOAP → farthest point sampling → reproducible rattling, one `.xyz` per material.

### Structures

```python
!mkdir -p data
!wget -q -O "data/Li10Ge(PS6)2_mp.cif" "https://raw.githubusercontent.com/dembart/intro-to-materials-informatics/refs/heads/main/data/seminar09/Li10Ge(PS6)2_mp.cif"

from ase.io import read, write

atoms = read("data/Li10Ge(PS6)2_mp.cif")
atoms.info["material_id"] = "lgps"
write("data/structures.xyz", [atoms])
```

### Settings

```python
!gridmlip dataset --print-config > dataset.yaml

import yaml

config = yaml.safe_load(open("dataset.yaml"))
config["specie"] = 3
config["defects"]["resolution"] = 0.4
config["selection"]["max_samples"] = 100
yaml.safe_dump(config, open("dataset.yaml", "w"), sort_keys=False)
```

### Dry run

```python
from gridmlip.dataset import build_dataset

build_dataset("data/structures.xyz", config="dataset.yaml", output_dir="datasets/lgps", dry_run=True)
```

### Build

```python
build_dataset("data/structures.xyz", config="dataset.yaml", output_dir="datasets/lgps")
```
