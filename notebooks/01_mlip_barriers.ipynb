# Percolation barriers with a batched MLIP

Li diffusion in Li10Ge(PS6)2 (LGPS) with a MACE foundation model.

### Structure

```python
!mkdir -p data
!wget -q -O "data/Li10Ge(PS6)2_mp.cif" "https://raw.githubusercontent.com/dembart/intro-to-materials-informatics/refs/heads/main/data/seminar09/Li10Ge(PS6)2_mp.cif"
```

### Grid

```python
from ase.io import read
import gridmlip as gm

atoms = read("data/Li10Ge(PS6)2_mp.cif")
grid = gm.Grid(atoms, specie="Li", resolution=0.2, r_min=0.8)
grid.summary()
```

### Energies (batched MACE)

```python
from gridmlip.predictors import mace

emap = grid.evaluate(mace("medium-mpa-0", device="auto"))
emap
```

### Percolation barriers

```python
emap.percolation_barriers()
```

### Save and visualize in VESTA

```python
emap.write("lgps_li.grd")
emap.save("lgps_li.npz")
```
